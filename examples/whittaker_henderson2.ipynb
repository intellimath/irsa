```python
tau2_slider = ipywidgets.FloatSlider(value=0.1, min=0, max=10, step=0.01, readout_format=".3f")
tau2_slider.layout.width="50%"
alpha_slider = ipywidgets.FloatSlider(value=0.01, min=0.0, max=1.0, step=0.01, readout_format=".3f")
alpha_slider.layout.width="50%"
xrange_slider = ipywidgets.FloatRangeSlider(
                value=(x.min(), x.max()),
                min=x.min(),
                max=x.max())
xrange_slider.layout.width="80%"
def GG(E):
    R = np.ones_like(E)
    mask = abs(inventory.modified_zscore(E)) > 3.0
    R[mask]=0
    return R

@ipywidgets.interact(tau2=tau2_slider, alpha=alpha_slider, xrange=xrange_slider, continuous_update=False)
def plot_smooth(tau2, alpha, xrange):
    yy, _ = smooth.whittaker_smooth_weight_func2(
            y_bl,
            func=GG,
            # func=funcs.RStep(1.0, eps=alpha),
            # func2=funcs.RStep(0.00, eps=0.001),
            # func2_e=inventory.relative_abs_max,
            tau2=tau2, d=2)
    err = y_bl - yy
    plt.figure(figsize=(12,4))
    plt.plot(x, y_bl, color='darkgray', linewidth=1.5)
    plt.plot(x, yy, color='r', linewidth=1.5)
    plt.plot(x, err , color='m', linewidth=1.5)
    plt.hlines(0, x.min(), x.max(), ls='--', lw=0.75)
    plt.tight_layout()
    plt.minorticks_on()
    plt.xlim(*xrange)
    plt.show()
```
